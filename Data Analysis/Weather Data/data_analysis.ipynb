```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('readings_complete.csv')
```

```python
df.shape
```

```python
station_stats = df.groupby(['region', 'station_id'])['temp'].agg(['mean', 'max', 'min', 'count']).reset_index()
station_stats['mean'] = station_stats['mean'].round(1)
print(station_stats)
```

```python
regional_weighted_avgs = station_stats.groupby('region')['mean'].mean()
print(regional_weighted_avgs)
```
